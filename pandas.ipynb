## Do movies added near Christmas have higher IMDb ratings?

Compare movies added from October 1 through December 29 (inclusive) with movies added during the rest of the year. This is a descriptive comparison, not proof that the add date caused a rating difference.

```python
movie_ratings = df_raw.loc[df_raw["type"].eq("movie"), ["title", "added_at", "imdb_rating"]].copy()
movie_ratings["added_at"] = pd.to_datetime(movie_ratings["added_at"], errors="coerce")
movie_ratings["imdb_rating"] = pd.to_numeric(movie_ratings["imdb_rating"], errors="coerce")

valid_ratings = movie_ratings.dropna(subset=["added_at", "imdb_rating"]).copy()
valid_ratings["near_christmas"] = (
    valid_ratings["added_at"].dt.month.ge(10)
    & ((valid_ratings["added_at"].dt.month.lt(12))
       | ((valid_ratings["added_at"].dt.month.eq(12)) & valid_ratings["added_at"].dt.day.le(29)))
)

rating_comparison = valid_ratings.groupby("near_christmas")["imdb_rating"].agg(
    movie_count="count", mean_rating="mean", median_rating="median"
).rename(index={False: "Rest of year", True: "Oct 1-Dec 29"})
display(rating_comparison.round(2))
print(f"Mean-rating difference (Oct 1-Dec 29 minus rest): "
      f"{valid_ratings.loc[valid_ratings['near_christmas'], 'imdb_rating'].mean() - valid_ratings.loc[~valid_ratings['near_christmas'], 'imdb_rating'].mean():.2f} points")
print(f"Movies excluded for a missing/unparseable add date or IMDb rating: "
      f"{len(movie_ratings) - len(valid_ratings)}")
```

## Regression: IMDb ratings and add date

Estimate the association between add date and IMDb rating for movies, controlling for release year, runtime, MPAA rating, and primary genre. The add-date coefficient is scaled to 30 days. Standard errors are clustered by add date because many titles share the same date; rare primary genres are grouped as Other. Rows missing the rating, date, release year, or runtime are excluded. This is an observational association, not a causal effect.

```python
import statsmodels.formula.api as smf

model_data = df_raw.loc[
    df_raw["type"].eq("movie"),
    ["imdb_rating", "added_at", "year", "runtime", "rated", "genre"]
].copy()
model_data["imdb_rating"] = pd.to_numeric(model_data["imdb_rating"], errors="coerce")
model_data["added_at"] = pd.to_datetime(model_data["added_at"], errors="coerce")
model_data["add_date_30d"] = (model_data["added_at"] - model_data["added_at"].min()).dt.days / 30
model_data["release_year"] = pd.to_numeric(
    model_data["year"].astype(str).str.extract(r"(\d{4})")[0], errors="coerce"
)
model_data["runtime_minutes"] = pd.to_numeric(
    model_data["runtime"].astype(str).str.extract(r"(\d+)")[0], errors="coerce"
)
model_data["rated_control"] = model_data["rated"].fillna("Unknown").replace({"N/A": "Unknown"})
model_data["primary_genre"] = model_data["genre"].fillna("Unknown").str.split(",").str[0].str.strip()
genre_counts = model_data["primary_genre"].value_counts()
model_data.loc[model_data["primary_genre"].map(genre_counts).lt(10), "primary_genre"] = "Other"
model_data = model_data.dropna(
    subset=["imdb_rating", "added_at", "release_year", "runtime_minutes"]
)

rating_model = smf.ols(
    "imdb_rating ~ add_date_30d + release_year + runtime_minutes "
    "+ C(rated_control) + C(primary_genre)",
    data=model_data
).fit(cov_type="cluster", cov_kwds={"groups": model_data["added_at"], "use_correction": True})

regression_table = pd.DataFrame({
    "Coefficient": rating_model.params,
    "Clustered SE": rating_model.bse,
    "95% CI lower": rating_model.conf_int()[0],
    "95% CI upper": rating_model.conf_int()[1],
    "p-value": rating_model.pvalues.map(lambda p: "<0.001" if p < 0.001 else f"{p:.3f}"),
})
regression_table.index = [
    ("Added date (per 30 days)" if term == "add_date_30d" else
     "Release year" if term == "release_year" else
     "Runtime (minutes)" if term == "runtime_minutes" else
     "Intercept" if term == "Intercept" else
     "MPAA rating: " + term.split("[T.", 1)[1].rstrip("]") if term.startswith("C(rated_control)[T.") else
     "Primary genre: " + term.split("[T.", 1)[1].rstrip("]"))
    for term in regression_table.index
]
display(regression_table.round({"Coefficient": 3, "Clustered SE": 3, "95% CI lower": 3, "95% CI upper": 3}))

model_fit = pd.DataFrame({
    "Metric": ["Movies in model", "Unique add dates", "R-squared", "Adjusted R-squared"],
    "Value": [int(rating_model.nobs), model_data["added_at"].nunique(),
              round(rating_model.rsquared, 3), round(rating_model.rsquared_adj, 3)],
})
display(model_fit)
```

## Regression: IMDb ratings and October-December release dates

This model compares movies released in October, November, or December with movies released January-September. IMDb rating is the outcome. Controls match the previous model: release year, runtime, MPAA rating, and primary genre. Standard errors are clustered by exact release date; rare primary genres are combined as Other. Rows missing a usable release date, IMDb rating, release year, or runtime are excluded. The Q4 coefficient is an adjusted association, not a causal effect.

```python
import statsmodels.formula.api as smf

release_model_data = df_raw.loc[
    df_raw["type"].eq("movie"),
    ["imdb_rating", "released_at", "year", "runtime", "rated", "genre"]
].copy()
release_model_data["imdb_rating"] = pd.to_numeric(
    release_model_data["imdb_rating"], errors="coerce"
)
release_model_data["released_at"] = pd.to_datetime(
    release_model_data["released_at"], format="%d %b %Y", errors="coerce"
)
release_model_data["release_year"] = pd.to_numeric(
    release_model_data["year"].astype(str).str.extract(r"(\d{4})")[0], errors="coerce"
)
release_model_data["runtime_minutes"] = pd.to_numeric(
    release_model_data["runtime"].astype(str).str.extract(r"(\d+)")[0], errors="coerce"
)
release_model_data["rated_control"] = (
    release_model_data["rated"].fillna("Unknown").replace({"N/A": "Unknown"})
)
release_model_data["primary_genre"] = (
    release_model_data["genre"].fillna("Unknown").str.split(",").str[0].str.strip()
)
release_genre_counts = release_model_data["primary_genre"].value_counts()
release_model_data.loc[
    release_model_data["primary_genre"].map(release_genre_counts).lt(10), "primary_genre"
] = "Other"
release_model_data["released_oct_dec"] = (
    release_model_data["released_at"].dt.month.ge(10).astype(int)
)
release_model_data = release_model_data.dropna(
    subset=["imdb_rating", "released_at", "release_year", "runtime_minutes"]
)

released_rating_model = smf.ols(
    "imdb_rating ~ released_oct_dec + release_year + runtime_minutes "
    "+ C(rated_control) + C(primary_genre)",
    data=release_model_data,
).fit(
    cov_type="cluster",
    cov_kwds={"groups": release_model_data["released_at"], "use_correction": True},
)

released_regression_table = pd.DataFrame({
    "Coefficient": released_rating_model.params,
    "Clustered SE": released_rating_model.bse,
    "95% CI lower": released_rating_model.conf_int()[0],
    "95% CI upper": released_rating_model.conf_int()[1],
    "p-value": released_rating_model.pvalues.map(
        lambda p: "<0.001" if p < 0.001 else f"{p:.3f}"
    ),
})
released_regression_table.index = [
    ("Released Oct-Dec (vs Jan-Sep)" if term == "released_oct_dec" else
     "Release year" if term == "release_year" else
     "Runtime (minutes)" if term == "runtime_minutes" else
     "Intercept" if term == "Intercept" else
     "MPAA rating: " + term.split("[T.", 1)[1].rstrip("]")
     if term.startswith("C(rated_control)[T.") else
     "Primary genre: " + term.split("[T.", 1)[1].rstrip("]"))
    for term in released_regression_table.index
]
main_release_terms = [
    "Released Oct-Dec (vs Jan-Sep)", "Release year", "Runtime (minutes)"
]
released_regression_table = released_regression_table.loc[
    main_release_terms + [
        term for term in released_regression_table.index if term not in main_release_terms
    ]
]
display(released_regression_table.round({
    "Coefficient": 3, "Clustered SE": 3,
    "95% CI lower": 3, "95% CI upper": 3,
}))

release_model_fit = pd.DataFrame({
    "Metric": ["Movies in model", "Unique release dates", "R-squared", "Adjusted R-squared"],
    "Value": [
        int(released_rating_model.nobs), release_model_data["released_at"].nunique(),
        round(released_rating_model.rsquared, 3),
        round(released_rating_model.rsquared_adj, 3),
    ],
})
display(release_model_fit)
```

## somewhat of a conclusion

My regression with the original release near christmas shows a positive assiciation with imdb ratings controlling for the included variables. Movies originally released near Christmas have higher IMDb ratings in this sample, whereas there is no statistically significant association between Christmas-period addition to Disney+ and ratings. The first regression is difficult to interpret seasonally because most titles were added on launch day.
